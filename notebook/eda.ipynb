# Retail Store Inventory

**Dataset Description** </br>

The dataset provides synthetic realistic data for analyzing and forecasting retail store inventory. 
It contains 73100 rows of daily data across multiple stores and products. This datasets have 15 columns like Date, Category, etc.
Column :
- Date : Daily records from [start_date] to [end_date]
- Store ID & Product ID : Unique identifiers for stores and products
- Category : Product categories like Groceries, Toys, Electronics, etc.
- Region : Geographic region of the store.
- Inventory Level : Stock available at the beginning of the day.
- Units Sold : Unit solds during the day.
- Units Ordered : Unit ordered during the day.
- Demand Forecast : Predicted demand based on past trends.
- Price : Price for the product.
- Discount : Discount for the product.
- Weather Condition : Daily Weather impacting sales.
- Holiday/Promotion : Indicators for holidays or promotions.
- Competitor Pricing : Price from competitor.
- Seasonality : Seasonality impacting sales.

## Exploratory Data Analytics

- Import library

```python
import pandas as pd
import plotly.express as px
import plotly.figure_factory as ff
import numpy as np
```

- Load data

```python
df = pd.read_csv('../data/raw/retail_store.csv')
df.head()
```

- Check info about data

```python
df.info()
```

- Check missing data

```python
df.isna().sum()
```

- Check duplicated data

```python
df.duplicated().sum()
```

- Change data type Data Column

```python
df['Date'] = df['Date'].apply(pd.to_datetime)
```

- Visualization Total Units Sold Over Time

```python
daily_sales = df.groupby("Date")["Units Sold"].sum().reset_index()
fig_daily = px.line(daily_sales, x='Date', y='Units Sold', title='Total Units Sold Over Time')

fig_daily.show()
```

- Visualization Sales Distribution by Day of Week

```python
df['day_of_week'] = df['Date'].dt.day_name()

fig_day_of_week = px.box(df, x='day_of_week', y='Units Sold', title='Sales Distribution by Day of Week')
fig_day_of_week.show()
```

- Visualization Impact of Promotion on Sales

```python
fig_impact = px.box(df, x="Holiday/Promotion", y="Units Sold", title="Impact of Promotion on Sales")
fig_impact.show()
```

- Visualization Price vs Units Sold

```python
fig_price_demand = px.scatter(df, x="Price", y="Units Sold", trendline="ols", title="Price vs Units Sold")
fig_price_demand.show()
```

- Visualization Correlation Heatmap

```python
corr = df.select_dtypes(include=['int64', 'float64']).corr()

fig_heatmap = ff.create_annotated_heatmap(
    z = np.round(corr.values, 6),
    x = list(corr.columns),
    y = list(corr.index),
    colorscale='Viridis'
)
fig_heatmap.show()
```

- Visualization Sales by Category

```python
category_sales = df.groupby("Category")["Units Sold"].sum().reset_index()

fig_sales = px.bar(category_sales, x="Category", y="Units Sold", title="Sales by Category")
fig_sales.show()
```
